## Parameter-overlap features

Accepted on two disjoint 2,452-query confirmation sets: Recall@50 0.781821 and 0.786416. The code below is the first confirmation driver; it uses existing E5 caches and never writes nswer.csv.


```python
"""Evaluate parameter-overlap features on untouched query groups; never writes answer.csv."""
from __future__ import annotations

import json
import time

import numpy as np
import pandas as pd

from audit import QUERY_COLUMNS
from dense_retrieval import embedding_cache, parameter_sets, scoped_dense, topk_dense
from geo_retrieval import LocationPrior, build_geo_channels
from retrieval_core import ROOT, confirmation_validation, fresh_validation, make_validation
from solution import ITEM_COLUMNS, prepare_items, predict, train_selector


def main():
    started = time.perf_counter()
    train = pd.read_parquet(ROOT / 'train.parquet',
                            columns=QUERY_COLUMNS + ITEM_COLUMNS + ['item_infm_params_text'])
    old, _, history = make_validation(train)
    fresh = fresh_validation(train, old, history)
    confirm = confirmation_validation(train, [old, fresh], history)
    queries = pd.concat([old, confirm], ignore_index=True)
    prior = LocationPrior(train.loc[train.query_group.isin(history)])
    items = prepare_items(train)
    items['dense_vector'] = list(embedding_cache(items, 'passage', 'train_e5_title_params'))
    queries['dense_vector'] = list(embedding_cache(queries, 'query', 'old_and_confirm_e5_queries'))
    channels, _, item_tokens, query_tokens = build_geo_channels(items, queries, prior)
    global_dense, global_scores = topk_dense(np.asarray(list(queries.dense_vector)),
                                             np.asarray(list(items.dense_vector)), k=1000)
    local_dense, local_scores = scoped_dense(queries, items, prior.scopes(queries, items), 200)
    channels = [(global_rows, local_rows, [[] for _ in range(len(queries))],
                 [[] for _ in range(len(queries))]) for global_rows, local_rows in channels]
    channels.append((global_dense, local_dense, global_scores, local_scores))
    params = parameter_sets(queries, items)
    positives = (train.loc[train.query_group.isin(set(queries.query_group))]
                 .groupby('query_group').item_id.agg(set).to_dict())
    model = train_selector(items, old, positives, channels, item_tokens, query_tokens,
                           prior=prior, dense=True, params=params)
    predictions = predict(model, items, queries, channels, item_tokens, query_tokens,
                          prior=prior, dense=True, params=params)[len(old):]
    values = [len(set(prediction) & positives[row.query_group]) / len(positives[row.query_group])
              for prediction, row in zip(predictions, confirm.itertuples(index=False))]
    report = {'parameter_hybrid_recall50': float(np.mean(values)), 'queries': len(values),
              'elapsed_seconds': time.perf_counter() - started}
    (ROOT / 'parameter_results.json').write_text(json.dumps(report, indent=2), encoding='utf-8')
    print(json.dumps(report, indent=2), flush=True)


if __name__ == '__main__':
    main()
```
